```python
from multi_llm_debate.analysis.analyze_convergence import analyze_round_number
from pathlib import Path

base_dir = Path("../data/bool_q")
df = analyze_round_number(
    base_dir=base_dir,
    max_round_number=10,
    cumulative=False,
    agent_count=3,
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
for idx, config in enumerate(df["model_configuration"]):
    rounds = range(1, 11)
    values = df.iloc[idx, 1:].values
    plt.plot(rounds, values, marker="o", label=config)

plt.xlabel("Round Number")
plt.ylabel("Number of Converged Debates")
plt.title("Convergence Analysis for Different 3-Agent Configurations")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
